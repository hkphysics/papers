# Integration of AI Large Language Models in High Energy Physics Research Workflows

## Abstract

Recent developments in artificial intelligence, particularly large language models (LLMs), have shown promise for improving productivity in high-energy physics. While there has been published work on the use of LLMs for data processing, very little has been published on their application to high-energy physics (HEP) research workflows.

In this paper, we describe our efforts to apply LLMs and AI agents to improve physics research workflows, with particular emphasis on how HEP research workflows differ from those of other fields, such as enterprise computing. Specifically, we describe our efforts to integrate LLMs into open-source computation engines such as Jupyter, and we describe how we have integrated tools such as OpenWebUI, LiteLLM, and Traefik into an integrated AI platform suitable for small research groups.

We then present our experiences with AI in research discovery, pedagogy, and data processing. Our key finding is that the specific model used is not important and that freely available models are more than sufficient for daily research work; the key challenges lie in integrating these tools into harnesses and other research infrastructure. We believe that improvements in research productivity do not require advances in the LLMs themselves, but rather improvements in coordinating harnesses and agents and integrating them into computational platforms. We conclude by discussing areas where we believe further research effort should be directed.
