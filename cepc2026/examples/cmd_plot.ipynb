```python
%matplotlib widget
import pandas as pd
import matplotlib.pyplot as plt
import urllib.request
import io
```

```python
csv_url = ("https://raw.githubusercontent.com/astronexus/HYG-Database/"
           "main/hyg/CURRENT/hygdata_v41.csv")
raw = urllib.request.urlopen(csv_url).read()
df = pd.read_csv(io.StringIO(raw.decode("utf-8")))
df = df.dropna(subset=["ci", "absmag"])
df = df[ (df["ci"] <= 2.0)]
plt.figure(figsize=(7, 8))
plt.scatter(df["ci"], df["absmag"], s=10, alpha=0.6)
plt.gca().invert_yaxis()
plt.xlabel("B-V color index")
plt.ylabel("Absolute visual magnitude M_V")
plt.title("Color-Magnitude Diagram (HYG v4.1)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
print(f"Stars plotted: {len(df)}")
```
